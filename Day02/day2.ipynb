### Fetching data from a REST API endpoint and converting it into a Python object using the requests library and json module.

```python
# Fetching data from a REST API endpoint and converting it into a Python object using the requests library and json module.

response= requests.get("https://api.github.com")

data = response.json()
print(data)

jsonData= json.dumps(data)
print(type(jsonData))
```
