## What is Good Prompt?
 A good prompt is a clear, specific instruction that guides an AI system to produce the exact output you need.
 System prompt + User prompt = output

### prompt format-
Role + Task + Context + Constraints 

#### Ex- 
Role:
You are a Python mentor.

Task:
Explain dictionaries.

Context:
I'm a third-year engineering student.

Constraints:
Use simple English and one real-life example.

```python
from dotenv import load_dotenv
import os
from google import genai
from google.genai import types

load_dotenv()

client =  genai.Client(api_key= os.getenv("GOOGLE_API_KEY"))




response= client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Review my resume",
    config=types.GenerateContentConfig(
        system_instruction="You are a senior HR recruiter."
    )
)

print(response.text)
```
